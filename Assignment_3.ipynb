```python
# Tic Tac Toe

# Board
board = [" "] * 9


# Display the board
def display():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]}")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]}")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]}")
    print()


# Check if a player has won
def check_winner(player):
    win = {
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    }

    for a, b, c in win:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


# AI move
def ai_move():

    # 1. Try to win
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"

            if check_winner("O"):
                return

            board[i] = " "

    # 2. Block the player
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"

            if check_winner("X"):
                board[i] = "O"
                return

            board[i] = " "

    # 3. Choose the first available position
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            return


# -------------------------
# Main Game Loop
# -------------------------

print("TIC TAC TOE")
print("You = X")
print("Computer = O")

while True:

    # Show board
    display()

    # -------------------------
    # User Move
    # -------------------------

    position = int(input("Enter position (1-9): ")) - 1

    # Check invalid position
    if position < 0 or position > 8 or board[position] != " ":
        print("Invalid move!")
        continue

    # Place X
    board[position] = "X"

    # Check if user won
    if check_winner("X"):
        display()
        print("You Win!")
        break

    # Check for draw
    if " " not in board:
        display()
        print("Draw!")
        break

    # -------------------------
    # AI Move
    # -------------------------

    ai_move()

    # Check if computer won
    if check_winner("O"):
        display()
        print("Computer Wins!")
        break

    # Check for draw
    if " " not in board:
        display()
        print("Draw!")
        break
```
