Matplotlib — Complete Basics
1. What is Matplotlib?
Matplotlib is a Python library used to create graphs and visualizations.

````python
import matplotlib.pyplot as plt
````

2. Basic Data
We'll use this data throughout the examples:

````python
students = ["Aryan", "Rahul", "Sneha", "Priya", "Amit"]

marks = [85, 78, 92, 88, 75]

age = [20, 21, 19, 22, 20]
````

3. Line Plot — plt.plot()
A line plot is useful for showing trends.

````python
plt.plot(students, marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
````

Add markers

````python
plt.plot(students, marks, marker="o")
plt.show()
````

Add a dashed line

````python
plt.plot(students, marks, linestyle="--")
plt.show()
````

Both Marker and Add a dashed line

````python
plt.plot(
    students,
    marks,
    marker="o",
    linestyle="--"
)

plt.show()
````

```
Common markers:
"o" -> circle
"s" -> square
"^" -> triangle
"*" -> star
"x" -> x

Common line styles:
"-"  -> solid
"--" -> dashed
":"  -> dotted
"-." -> dash-dot
```

4. Bar Chart — plt.bar()

5. Scatter Plot — plt.scatter()
Used to show the relationship between two numerical variables.

````python
plt.scatter(age, marks)

plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks")

plt.show()
````

6. Pie Chart — plt.pie()
Used to show parts/proportions of a whole.

````python
subjects = ["Maths", "Python", "DBMS", "Java"]
subject_marks = [85, 90, 30, 80]

plt.pie(
    subject_marks,
    labels=subjects
)

plt.title("Subject Marks")

plt.show()
````

7. Histogram — plt.hist()
Used to show the distribution of numerical data.

````python
marks_data = [
    45, 55, 60, 62, 65,
    68, 70, 72, 75, 78,
    80, 82, 85, 88, 90,
    92, 95
]

plt.hist(marks_data)

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")

plt.show()
````

Using bins

````python
plt.hist(marks_data, bins=5)

plt.show()
````

13. Subplot — plt.subplot()
Allows you to put multiple plots in one figure.

````python
plt.figure(figsize=(10, 8))  # Adding figsize also gives subplots more breathing room

plt.subplot(2, 2, 1)
plt.plot(students, marks)
plt.title("Line Plot")

plt.subplot(2, 2, 2)
plt.bar(students, marks)
plt.title("Bar Chart")

plt.subplot(2, 2, 3)
plt.scatter(age, marks)
plt.title("Scatter Plot")

plt.subplot(2, 2, 4)
plt.hist(marks)
plt.title("Histogram")

plt.tight_layout()  # Fixes the overlapping labels and titles
plt.show()
````

```
The syntax is:
plt.subplot(rows, columns, position)


So:
plt.subplot(2, 2, 1)


means:
2 rows
2 columns
position 1

Layout:
┌────────────┬────────────┐
│ Plot 1     │ Plot 2     │
├────────────┼────────────┤
│ Plot 3     │ Plot 4     │
└────────────┴────────────┘
```

14. plt.subplots()
This is another, often better, way to create multiple plots.

````python
fig, ax = plt.subplots(2, 2)
````

````python
import matplotlib.pyplot as plt

students = ["Aryan", "Rahul", "Sneha", "Priya", "Amit"]
marks = [85, 78, 92, 88, 75]
age = [20, 21, 19, 22, 20]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# 1. Line Plot
ax[0, 0].plot(students, marks, marker="o")
ax[0, 0].set_title("Line Plot")
ax[0, 0].set_xlabel("Students")
ax[0, 0].set_ylabel("Marks")

# 2. Bar Chart
ax[0, 1].bar(students, marks)
ax[0, 1].set_title("Bar Chart")
ax[0, 1].set_xlabel("Students")
ax[0, 1].set_ylabel("Marks")

# 3. Scatter Plot
ax[1, 0].scatter(age, marks)
ax[1, 0].set_title("Scatter Plot")
ax[1, 0].set_xlabel("Age")
ax[1, 0].set_ylabel("Marks")

# 4. Histogram
ax[1, 1].hist(marks, bins=5)
ax[1, 1].set_title("Histogram")
ax[1, 1].set_xlabel("Marks")
ax[1, 1].set_ylabel("Frequency")

# Adjust spacing
plt.tight_layout()

# Display
plt.show()
````

````python
ax[0, 0].set_title("Line Plot")
ax[0, 1].set_title("Bar Chart")
ax[1, 0].set_title("Scatter Plot")
ax[1, 1].set_title("Histogram")
````


26. Complete subplot() example

This is a very good example for your practical:

````python
import matplotlib.pyplot as plt

students = ["Aryan", "Rahul", "Sneha", "Priya", "Amit"]

marks = [85, 78, 92, 88, 75]

age = [20, 21, 19, 22, 20]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# Line Plot
ax[0, 0].plot(students, marks, marker="o")
ax[0, 0].set_title("Line Plot")
ax[0, 0].set_xlabel("Students")
ax[0, 0].set_ylabel("Marks")

# Bar Chart
ax[0, 1].bar(students, marks)
ax[0, 1].set_title("Bar Chart")
ax[0, 1].set_xlabel("Students")
ax[0, 1].set_ylabel("Marks")

# Scatter Plot
ax[1, 0].scatter(age, marks)
ax[1, 0].set_title("Age vs Marks")
ax[1, 0].set_xlabel("Age")
ax[1, 0].set_ylabel("Marks")

# Histogram
ax[1, 1].hist(marks, bins=5)
ax[1, 1].set_title("Marks Distribution")
ax[1, 1].set_xlabel("Marks")
ax[1, 1].set_ylabel("Frequency")

plt.tight_layout()

plt.show()
````
