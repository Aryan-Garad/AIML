```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("titanic.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Missing values before preprocessing:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Fill Age using median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill Embarked using mode
df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)

# Drop Cabin due to excessive missing values
df.drop(columns=["Cabin"], inplace=True)

print("Missing values after preprocessing:")
print(df.isnull().sum())

print("Total remaining missing values:", df.isnull().sum().sum())
```

```python
y = df["Survived"]

X = df.drop(
    columns=["Survived", "PassengerId", "Name", "Ticket"]
)

print("Input features:")
print(X.columns.tolist())

print("\nTarget variable: Survived")
```

```python
X = pd.get_dummies(
    X,
    columns=["Pclass", "Sex", "Embarked"],
    dtype=int
)

display(X.head())
```

```python
numeric_features = ["Age", "SibSp", "Parch", "Fare"]

scaler = MinMaxScaler()

X[numeric_features] = scaler.fit_transform(
    X[numeric_features]
)

display(X.head())
```

```python
processed = X.copy()
processed["Survived"] = y

processed.to_csv(
    "titanic_preprocessed_encoded_normalized.csv",
    index=False
)

print("Preprocessing completed successfully!")
print("Final dataset shape:", processed.shape)
print("Missing values:", processed.isnull().sum().sum())

display(processed.head())
```
