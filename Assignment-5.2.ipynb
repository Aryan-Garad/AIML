```python
import matplotlib.pyplot as plt
```

```python
students = ["Aryan", "Rahul", "Sneha", "Priya", "Amit"]
marks = [85, 78, 92, 88, 75]
```

```python
plt.plot(students, marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

```python
plt.bar(students, marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

```python
age = [20, 21, 19, 22, 20]
marks = [85, 78, 92, 88, 75]

plt.scatter(age, marks)

plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks")

plt.show()
```

```python
subjects = ["Maths", "Python", "DBMS", "Java"]
marks = [85, 90, 75, 80]

plt.pie(marks, labels=subjects, autopct="%1.1f%%")

plt.title("Subject Marks Distribution")

plt.show()
```

```python
marks = [45, 55, 60, 62, 65, 68, 70, 72, 75, 78,
         80, 82, 85, 88, 90, 92, 95]

plt.hist(marks, bins=5)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Aryan", "Rahul", "Sneha", "Priya", "Amit"]
marks = [85, 78, 92, 88, 75]

plt.plot(students, marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.grid()

plt.show()
```

```python
students = ["Aryan", "Rahul", "Sneha", "Priya", "Amit"]
marks = [85, 78, 92, 88, 75]
```

```python
import matplotlib.pyplot as plt

plt.plot(students, marks, marker="o", linestyle="--")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.grid()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Aryan", "Rahul", "Sneha", "Priya", "Amit"],
    "Marks": [85, 78, 92, 88, 75]
}

df = pd.DataFrame(data)

plt.bar(df["Name"], df["Marks"])

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```
