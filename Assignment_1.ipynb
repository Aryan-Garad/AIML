```python
# dfs using stack
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

visited = []
stack = []

def dfs(node):
    stack.append(node)

    while stack:
        s = stack.pop()

        if s not in visited:
            print(s, end=" ")
            visited.append(s)

            # Push neighbors in reverse order
            for neighbor in reversed(graph[s]):
                if neighbor not in visited:
                    stack.append(neighbor)

start = input("Enter the starting node: ").strip().upper()

print("DFS Traversal:")
dfs(start)
```

```python
# dfs using recursion
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

visited = set()

def dfs(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbor in graph[node]:
            dfs(neighbor)

start = input("Enter the starting node: ").upper()

print("DFS Traversal:")
dfs(start)
```

```python
#bfs
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

visited = []
queue = []

def bfs(node):
    visited.append(node)
    queue.append(node)

    while queue:
        s = queue.pop(0)
        print(s, end=" ")

        for neighbor in graph[s]:
            if neighbor not in visited:
                visited.append(neighbor)
                queue.append(neighbor)

start = input("Enter the starting node: ").strip().upper()

print("BFS Traversal:")
bfs(start)
```
